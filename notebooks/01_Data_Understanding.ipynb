The dataset contains three accident severity categories: Slight Injury, Serious Injury, and Fatal Injury. Among 8,210 accident records, 7,082 are classified as Slight Injury, 1,046 as Serious Injury, and only 82 as Fatal Injury. This shows that most accidents belong to the Slight Injury category, while Fatal Injury is the least common. The dataset is imbalanced because the number of records differs greatly between the three categories.
